We take the 52-page PDF, already split into 227 numbered clauses.

```python
from labcore import corpus

chunks, texts, manifest = corpus()

print(f"{len(chunks)} chunks from {manifest['source_pdf']}")
print(f"built with: {manifest['embedding_model']}")
```

## What does one chunk look like?

```python
print(texts[0][:400])
```

Notice the `[General]` prefix. **Every chunk starts with its section name.**

That is not decoration — it is what makes a search for "attendance" find the attendance clauses at all. Without it, retrieval failed completely in testing.

## Find the clauses that matter

```python
# The attendance rules - the ones this lab keeps coming back to
for c in chunks:
    if c["section"].lower().startswith("attend"):
        print(f"p{c['page']}  {c['text'][:110]}...")
        print()
```

---

## Now change it yourself

Change `"attend"` below to explore other parts of your regulations.

```python
# Try: "assess", "award", "grade", "curriculum", "change of branch"
want = "award"

hits = [c for c in chunks if want.lower() in c["section"].lower()]
print(f"{len(hits)} chunks in sections matching '{want}'\n")
for c in hits[:5]:
    print(f"p{c['page']}  [{c['section']}]")
    print(f"      {c['text'][:150]}...")
    print()
```
